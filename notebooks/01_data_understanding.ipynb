## Inspect CSV Files

This section previews the downloaded CSV files so that I can understand their column structure, row counts and suitability for analysis.

```python
# Inspect CSV files

csv_files = [file for file in raw_files if file.suffix.lower() == ".csv"]

print("CSV files found:", len(csv_files))

for file in csv_files:
    print("\n" + "=" * 80)
    print("File:", file.name)
    print("=" * 80)
    
    try:
        df_preview = pd.read_csv(file, nrows=5, low_memory=False)
        df_full_sample = pd.read_csv(file, nrows=1000, low_memory=False)
        
        print("Preview shape:", df_preview.shape)
        print("Columns:")
        print(df_preview.columns.tolist())
        display(df_preview.head())
        
    except Exception as error:
        print("Could not read file:", file.name)
        print("Error:", error)
```

## Inspect ODS Road Freight Files

This section checks the ODS files downloaded from GOV.UK road freight statistics tables.

ODS files can contain multiple sheets and may have title/header rows before the actual data table.

```python
# Inspect ODS files

ods_files = [file for file in raw_files if file.suffix.lower() == ".ods"]

print("ODS files found:", len(ods_files))

for file in ods_files:
    print("\n" + "=" * 80)
    print("File:", file.name)
    print("=" * 80)
    
    try:
        excel_file = pd.ExcelFile(file, engine="odf")
        print("Sheet names:")
        print(excel_file.sheet_names)
        
    except Exception as error:
        print("Could not open ODS file:", file.name)
        print("Error:", error)
```

```python
# Preview the first sheet from each ODS file

for file in ods_files:
    print("\n" + "=" * 80)
    print("File:", file.name)
    print("=" * 80)
    
    try:
        excel_file = pd.ExcelFile(file, engine="odf")
        first_sheet = excel_file.sheet_names[0]
        
        print("Previewing sheet:", first_sheet)
        
        df_ods_preview = pd.read_excel(
            file,
            sheet_name=first_sheet,
            engine="odf",
            header=None,
            nrows=15
        )
        
        display(df_ods_preview)
        
    except Exception as error:
        print("Could not preview ODS file:", file.name)
        print("Error:", error)
```

```python
# Create ODS sheet inventory

ods_sheet_inventory = []

for file in ods_files:
    try:
        excel_file = pd.ExcelFile(file, engine="odf")
        
        for sheet_name in excel_file.sheet_names:
            ods_sheet_inventory.append({
                "file_name": file.name,
                "sheet_name": sheet_name
            })
            
    except Exception as error:
        ods_sheet_inventory.append({
            "file_name": file.name,
            "sheet_name": f"ERROR: {error}"
        })

ods_sheet_inventory_df = pd.DataFrame(ods_sheet_inventory)

ods_sheet_inventory_df
```
